```python
df = pd.read_csv('gurgaon_properties_13k_missing_value_imputation.csv')
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
def categorize_luxury(score):
    if 0 <= score < 50:
        return "Low"
    elif 50 <= score < 150:
        return "Medium"
    elif score >= 150:
        return "High"
    else:
        return None
```

```python
df['luxury_category'] = df['luxury_score'].apply(categorize_luxury)
```

```python
def categorize_floor(floor):
    if 0 <= floor <= 2:
        return "Low Floor"
    elif 3 <= floor <= 10:
        return "Mid Floor"
    elif 11 <= floor <= 51:
        return "High Floor"
    else:
        return None
```

```python
df['floor_category'] = df['floorNum'].apply(categorize_floor)
```

```python
df['furnishing_type'].value_counts()
```

```python
## 0 -> Unfurnished
## 1 -> Semifurnished
## 2 -> Furnished

df['furnishing_type'] = df['furnishing_type'].replace({0: 'Unfurnished', 1: 'Semifurnished', 2: 'Furnished'})
```

```python
df["sector"] = df["sector"].str.capitalize()
```

```python
df.drop(columns= ['pooja room', 'study room', 'others', 'floorNum', 'luxury_score', 'society', 'price_per_sqft'], inplace= True)
```
